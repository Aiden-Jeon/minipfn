노트북을 시작하기 전에 패키지와 연산 장치가 제대로 잡히는지 확인합니다. 로컬(uv)과 Databricks Serverless GPU 양쪽에서 그대로 실행되도록 작성했습니다.

## `minipfn` 불러오기

로컬에서는 `uv sync`로 패키지가 설치되어 있습니다. Databricks처럼 패키지가 설치되지 않은 환경에서는 저장소의 `src/`를 import 경로에 추가합니다. 모든 챕터 노트북은 이 셀로 시작합니다.

```python
try:
    import minipfn
except ImportError:
    import sys
    from pathlib import Path

    sys.path.insert(0, str(Path.cwd().parent / "src"))
    import minipfn

print(minipfn.__file__)
```

```python
import sys

import matplotlib
import numpy as np
import scipy
import sklearn
import torch

from minipfn import get_device, set_seed

print(f"python       {sys.version.split()[0]}")
for mod in (torch, np, scipy, sklearn, matplotlib):
    print(f"{mod.__name__:<12} {mod.__version__}")
```

## 연산 장치

`get_device()`는 cuda > mps > cpu 순서로 쓸 수 있는 장치를 고릅니다. Databricks Serverless GPU(A10 / H100)에서는 `cuda`, Apple Silicon 로컬에서는 `mps`가 나와야 합니다.

```python
set_seed(42)
device = get_device()
print("device:", device)
if device.type == "cuda":
    print("cuda   :", torch.version.cuda)
    print("gpu    :", torch.cuda.get_device_name(0))

x = torch.randn(1024, 1024, device=device)
print((x @ x.T).shape)
```
